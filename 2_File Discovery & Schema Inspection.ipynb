```python
import os
import glob
import pandas as pd
import numpy as np

# 1. Define directory path (using raw string 'r' to avoid Windows backslash errors)
DATA_DIR = r"C:\Users\admin\OneDrive\Desktop\Pro-1\Parquet"

parquet_files = sorted(glob.glob(os.path.join(DATA_DIR, "*.parquet")))
print(f"Found {len(parquet_files)} parquet files:\n")

orig_files = []
perf_files = []

for f in parquet_files:
    fname = os.path.basename(f)
    # Read just 2 rows to inspect schema without using RAM
    sample_df = pd.read_parquet(f).head(2)
    num_cols = sample_df.shape[1]
    
    # Freddie Mac Origination has ~32 columns; Performance (svcg) has ~32 columns
    # Performance files usually have 'svcg' or 'time' in the filename
    if "svcg" in fname.lower() or "time" in fname.lower() or "perf" in fname.lower():
        perf_files.append(f)
        ftype = "PERFORMANCE (Monthly)"
    else:
        orig_files.append(f)
        ftype = "ORIGINATION (Static)"
        
    print(f"[{ftype}] {fname} | Columns: {num_cols}")
    print(f"   Sample Columns: {list(sample_df.columns[:6])}\n")
```
